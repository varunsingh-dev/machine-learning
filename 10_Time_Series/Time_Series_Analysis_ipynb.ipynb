```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

##  Load Dataset

```python
df = pd.read_csv("/content/sales_data.csv")

df.head()
```

##  Dataset Shape

```python
df.shape
```

##  Dataset Columns

```python
df.columns.tolist()
```

##  Dataset Information

```python
df.info()
```

```python
df.columns.tolist()
```

##  Check Missing Values

```python
df.isnull().sum()
```

##  Convert Date Column

```python
df["Date"] = pd.to_datetime(df["Date"])

df.head()
```

##  Check Date Range

```python
print("Start Date:", df["Date"].min())
print("End Date:", df["Date"].max())
```

##  Sort Data by Date

```python
df = df.sort_values("Date")

df.head()
```

##  Set Date as Index

```python
df = df.set_index("Date")

df.head()
```

##  Sales Over Time

```python
plt.figure(figsize=(12, 6))

plt.plot(df.index, df["Revenue"])

plt.title("Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")

plt.grid(True)
plt.show()
```

##  Revenue Statistics

```python
print("Maximum Revenue:", df["Revenue"].max())
print("Minimum Revenue:", df["Revenue"].min())
print("Average Revenue:", df["Revenue"].mean())
print("Total Revenue:", df["Revenue"].sum())
```

##  Moving Average Analysis

```python
df["Moving_Average"] = df["Revenue"].rolling(window=7).mean()

df[["Revenue", "Moving_Average"]].head(10)
```

##  Revenue and Moving Average

```python
plt.figure(figsize=(12, 6))

plt.plot(df.index, df["Revenue"], label="Revenue")
plt.plot(
    df.index,
    df["Moving_Average"],
    label="7-Day Moving Average"
)

plt.title("Revenue and Moving Average Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)

plt.show()
```

##  Monthly Revenue Analysis

```python
monthly_revenue = df.resample("ME")["Revenue"].sum()

monthly_revenue.head()
```

##  Monthly Revenue Trend

```python
plt.figure(figsize=(12, 6))

plt.plot(monthly_revenue.index, monthly_revenue.values, marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")

plt.grid(True)
plt.show()
```

##  Yearly Revenue Analysis

```python
yearly_revenue = df.resample("YE")["Revenue"].sum()

yearly_revenue
```

##  Daily Revenue Analysis

```python
daily_revenue = df["Revenue"].resample("D").sum()

daily_revenue.head(10)
```

##  Daily Revenue Trend

```python
plt.figure(figsize=(12, 6))

plt.plot(daily_revenue.index, daily_revenue.values)

plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Total Revenue")

plt.grid(True)
plt.show()
```

##  30-Day Moving Average

```python
daily_moving_average = daily_revenue.rolling(window=30).mean()

daily_moving_average.head(35)
```

##  Revenue Trend with Moving Average

```python
plt.figure(figsize=(12, 6))

plt.plot(
    daily_revenue.index,
    daily_revenue.values,
    label="Daily Revenue"
)

plt.plot(
    daily_moving_average.index,
    daily_moving_average.values,
    label="30-Day Moving Average"
)

plt.title("Daily Revenue and 30-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)

plt.show()
```

##  Yearly Revenue Trend

```python
plt.figure(figsize=(10, 5))

plt.plot(
    yearly_revenue.index,
    yearly_revenue.values,
    marker="o"
)

plt.title("Yearly Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Total Revenue")

plt.grid(True)
plt.show()
```

##  Best Revenue Month

```python
best_month = monthly_revenue.idxmax()
best_month_revenue = monthly_revenue.max()

print("Best Revenue Month:", best_month)
print("Revenue:", best_month_revenue)
```

##  Best Revenue Year

```python
best_year = yearly_revenue.idxmax()
best_year_revenue = yearly_revenue.max()

print("Best Revenue Year:", best_year.year)
print("Revenue:", best_year_revenue)
```
